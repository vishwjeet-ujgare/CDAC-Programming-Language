```python
%%writefile matrix_add.c

#include <stdio.h>
#include <openacc.h>

#define ROWS	3
#define	COLS	3

int main() {

	printf("Starting ...\n");


  // Initialize matrices
  int A[ROWS][COLS];
  int B[ROWS][COLS];
  int C[ROWS][COLS];

  for (int i = 0; i < ROWS; ++i) {
    for (int j = 0; j < COLS; ++j) {
      A[i][j] = 1;
      B[i][j] = 2;
    }
  }

  // Matrix addition
  //#pragma acc kernels
  for (int i = 0; i < ROWS; ++i) {
    for (int j = 0; j < COLS; ++j) {
      C[i][j] = A[i][j] + B[i][j];
    }
  }

  // Print result matrix
  printf("Resultant Matrix:\n");
  for (int i = 0; i < ROWS; ++i) {
    for (int j = 0; j < COLS; ++j) {
      printf("%d ", C[i][j]);
    }
    printf("\n");
  }

  return 0;
}
```
